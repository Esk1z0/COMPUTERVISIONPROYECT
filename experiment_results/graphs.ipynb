```python
import matplotlib.pyplot as plt
```

```python
names = [
    'Basic CNN',
    'Data Augmentation',
    'Classification Layer',
    'Reescaling',
    'Dropout 0.8',
    'Batch Normalization',
]
metrics = [
    0.465669,
    0.437039,
    0.452144,
    0.632220,
    0.698301,
    0.569067,
]
```

```python
dropout_nets = [
    'Dropout 0.1',
    'Dropout 0.3',
    'Dropout 0.5',
    'Dropout 0.6',
    'Dropout 0.7',
    'Dropout 0.8',
    'Dropout 0.9',
]
metrics_dropout = [
    0.644282,
    0.670409,
    0.657933,
    0.683139,
    0.660628,
    0.698301,
    0.664472,
]
```

```python
def plot_points(x, y):
    plt.figure(figsize=(10, 6))
    plt.plot(x, y, marker='o')
    plt.title('Model Performance')
    plt.xlabel('Model Variations')
    plt.ylabel('Accuracy')
    plt.xticks(rotation=45)
    plt.grid()
    plt.show()
```

```python
plot_points(names, metrics)
```

```python
plot_points(dropout_nets, metrics_dropout)
```
